# Phishing URL detection: local Jupyter runbook
Run the cells **in order** (Shift+Enter). Project folder: `Downloads/phishing_project`.
Pick the profile in Step 2: `laptop` (fast, uses a subsample of the training rows) or `full` (all training rows, long on a CPU).

## Step 1. Set up the project folder and a helper that runs the pipeline

```python
from pathlib import Path
import os, sys, subprocess, shlex, json

PROJ = Path.home() / 'Downloads' / 'phishing_project'      # change this if your folder is somewhere else
assert (PROJ / 'phishing_pipeline.py').exists(), f'phishing_pipeline.py not found in {PROJ}'
os.chdir(PROJ)
(PROJ / 'data').mkdir(exist_ok=True)
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

def run(args):
    """Run `python phishing_pipeline.py <args>` with this notebook's Python and stream the output."""
    cmd = [sys.executable, '-u', 'phishing_pipeline.py'] + shlex.split(args)
    p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
    for line in p.stdout:
        print(line, end='')
    p.wait()
    print('exit code:', p.returncode)

print(sys.version)
import tensorflow as tf
print('TensorFlow', tf.__version__, '| GPU:', tf.config.list_physical_devices('GPU') or 'none (CPU only)')
```

## Step 2. Choose the profile
* `laptop`: 40,000 training rows, 2,000 text features, 8 epochs, no hyper-parameter search. Use this on a normal laptop CPU.
  **If you use it, the paper must say that training used a 40,000-URL subsample** (validation and test sets are always complete).
* `full`: all 188k training rows, 5,000 features, 20 epochs, hyper-parameter search. Needs a strong PC or a GPU.

```python
PROFILE = 'laptop'          # 'laptop' or 'full'
if PROFILE == 'laptop':
    EXTRA, EPOCHS, TUNE = '--max-train-rows 40000 --max-features 2000', 8, ''
else:
    EXTRA, EPOCHS, TUNE = '', 20, '--tune'
OUT = f'results_{PROFILE}'
print(PROFILE, '->', OUT, '|', EXTRA, '| epochs', EPOCHS, '|', TUNE or 'no tuning')
```

## Step 3. Download the PhiUSIIL dataset (UCI) into `data/`

```python
import urllib.request, zipfile
URL = 'https://archive.ics.uci.edu/static/public/967/phiusiil+phishing+url+dataset.zip'
zpath = PROJ / 'data' / 'phiusiil.zip'
if not (PROJ / 'data' / 'PhiUSIIL_Phishing_URL_Dataset.csv').exists():
    req = urllib.request.Request(URL, headers={'User-Agent': 'Mozilla/5.0'})
    with urllib.request.urlopen(req) as r, open(zpath, 'wb') as f:
        f.write(r.read())
    zipfile.ZipFile(zpath).extractall(PROJ / 'data')
print(os.listdir(PROJ / 'data'))
```

If the download fails (network or firewall), open https://archive.ics.uci.edu/dataset/967/phiusiil+phishing+url+dataset in a browser, click **Download**, unzip it, and put `PhiUSIIL_Phishing_URL_Dataset.csv` into the `data` folder. Then run the next cell.

```python
import pandas as pd
df = pd.read_csv('data/PhiUSIIL_Phishing_URL_Dataset.csv', usecols=['URL', 'label'])
print(df.shape)
print(df.label.value_counts())      # expected: 1 -> 134,850 (legitimate), 0 -> 100,945 (phishing); the script flips this
```
